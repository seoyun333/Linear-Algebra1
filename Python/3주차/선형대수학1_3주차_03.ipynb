```python
#3주차_03 정방행렬 대각합 구하기1

#행렬
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

#대각 합 계산
diagonal_sum = sum(matrix[i][i] for i in range(len(matrix)))

#결과 출력
print("대각합: ", diagonal_sum)
```
