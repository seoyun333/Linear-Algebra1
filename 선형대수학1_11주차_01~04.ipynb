```python
# 실습 01

import numpy as np

A = np.array([
    [2, 3],
    [3, -6]
], dtype = float)

# 특성다항식 계산
poly = np.poly(A)

# 고유값 계산
eigenvalues = np.linalg.eigvals(A)

# MATLAB 화면처럼 고유값을 -7, 3 순서로 정렬
eigenvalues = np.sort(eigenvalues)

# 소수점 없이 정수로 변환
poly = np.rint(poly).astype(int)
eigenvalues = np.rint(eigenvalues).astype(int)

print("특성다항식 =")
print(poly)
print("\n고유값 =")
print(eigenvalues)
```

```python
# 실습 02

import numpy as np

A = np.array([
    [3, 2],
    [3, -2]
], dtype = float)

# 고유값과 고유벡터 계산
eigenvalues, X = np.linalg.eig(A)

# MATLAB 화면처럼 고유값을 4, -3 순서로 정렬
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
X = X[:, order]

# 고유값을 대각행렬로 변환하고 정수로 변환
lamda = np.rint(np.diag(eigenvalues)).astype(int)

np.set_printoptions(precision=4, suppress=True, floatmode='fixed')

print("X =")
print(X)
print("\nlamda =")
print(lamda)
```

```python
# 실습 03

import numpy as np

A = np.array([
    [3, -1, -1],
    [-12, 0, 5],
    [4, -2, -1]
], dtype = float)

# 고유값과 고유벡터 계산
eigenvalues, X = np.linalg.eig(A)

# MATLAB 화면처럼 고유값을 2, 1, -1 순서로 정렬
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
X = X[:, order]

# 고유값을 대각행렬로 변환
lamda = np.diag(eigenvalues)

np.set_printoptions(precision=4, suppress=True, floatmode='fixed')

print("X =")
print(X)
print("\nlamda =")
print(lamda)
```

```python
# 실습 04

import numpy as np

A = np.array([
    [0, 0, 0],
    [0, 2, 5],
    [0, 0, -1]
], dtype = float)

B = np.array([
    [4, 0, 0],
    [0, 0, 0],
    [1, 0, -3]
], dtype = float)

# 행렬 A, B의 고유값 계산
eigenvalues_A = np.linalg.eigvals(A)
eigenvalues_B = np.linalg.eigvals(B)

# 고유값을 소수점 없이 정수로 변환
eigenvalues_A = np.rint(eigenvalues_A).astype(int)
eigenvalues_B = np.rint(eigenvalues_B).astype(int)

print("A의 고유값 =")
print(eigenvalues_A)
print("\nB의 고유값 =")
print(eigenvalues_B)
```
